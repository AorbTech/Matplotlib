### What is a bar chart?

- A bar chart is used when we want to compare categorical, or grouped, data.
- It uses rectangular bars with height proportional to the data they represent.
- Commonly used for comparisons and rankings.

![cover](Graph/2.%20Bar%20Chart.png)

### Import Matplalib

```python
import matplotlib.pyplot as plt
```

### Import Dataset

```python
# House Price Averages by Year (eurostat)
years = [2014, 2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022, 2023, 2024]
avg_house_price = [97.8, 100, 104.3, 109.3, 114.8, 120.4, 127.1, 137.8, 148.9, 148.4, 153.4]
# Ice Cream Flavour Popularity
flavours = ["Vanilla", "Chocolate", "Strawberry", "Pistachio", "Matcha-Strawberry", "Cookies n' Cream"]
votes = [87, 112, 64, 45, 58, 93]
```

### Bar Chart

```python
plt.bar(years, avg_house_price)
```

### Customizing the Bar Chart

```python
plt.bar(years, avg_house_price, label='Price', color= 'yellowgreen')
plt.title('House Price Averages by Year')
plt.xlabel('Year')
plt.ylabel('Average House Price (in thousands)')
plt.legend()
plt.grid(True)
plt.ylim(95, 155)
plt.show()
```

### Horizontal Bar Chart
By using `barh` we can make the bar chart horizontal

```python
plt.barh(flavours, votes, label='No. votes')
plt.title('Ice Cream Flavour Popularity')
plt.xlabel('Number of Votes')
plt.ylabel('Flavours')
plt.legend()
plt.grid(True)
plt.show()
```

### Plot Graph with Different Themes

```python
print(plt.style.available)
```

```python
# solarizedlight style
plt.style.use('Solarize_Light2')

plt.barh(flavours, votes, label='No. votes', color= 'pink')
plt.title('Ice Cream Flavour Popularity')
plt.xlabel('Number of Votes')
plt.ylabel('Flavours')
plt.legend()
plt.grid(True)
plt.show()
```

```python
# dark background style
plt.style.use('dark_background')

plt.barh(flavours, votes, label='No. votes', color= 'pink')
plt.title('Ice Cream Flavour Popularity')
plt.xlabel('Number of Votes')
plt.ylabel('Flavours')
plt.legend()
plt.grid(True)
plt.show()
```
