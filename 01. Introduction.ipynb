### What is Matplotlib?
- Matplotlib is a Python library used for data visualization.
- It helps you create plots, charts, and graphs


### Founder of Matplotlib

- Founder: John D. Hunter
- Year: 2002
- He created Matplotlib because Python lacked good graph-drawing tools
- Wanted a free alternative to MATLAB
- Designed it for easy data visualization
- Later maintained by the open-source community

### Features of Matplotlib

- Open-source and free
- Supports 2D and limited 3D plotting
- Highly customizable plots
- Works with NumPy, Pandas, and SciPy
- Supports multiple output formats (PNG, PDF, SVG, etc.)
- Cross-platform (Windows, Linux, macOS)

### Applications of Matplotlib
- Statistical Data analysis and visualization
- Machine learning model evaluation
- Scientific research and reports
